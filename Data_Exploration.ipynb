# Data exploration Notebook

The purpose of this notebook is to explore the datasets of interests to use for this DIS study project. As based on the accompanyng topic paper, this study is an exploration to predict levels of DNA methylation potentially related to diseaes via the nature of such methylation associated with urban features and landscapes associated and classified under concepts of New Urbaanism. In simplier terms, the project is focused an input X and output Y prediction model. X inputs are the landscape features (new urbanism) while Y is the methylation score (bioinformatics). To make sure that there are no discrepancies for further analysis, the datasets needs to be explored, especially regarding the zip code data, which serves as a link between the inputs X and output Y. After examining the output databses, a decision can be made if requested permission needs to be made to access more information, particiular if Zip code information can be provided.

There are four datasets of interest to explore

### Input X databases
- **EPA Smart Location Mapping Database (SLD)**
- **CDC Places Database**

### Output Y databases 
- **CDC National Health and Nutrition Examination Survey (NHNES) Database**
- **NIH All of Us Research Program (AOUP)**
