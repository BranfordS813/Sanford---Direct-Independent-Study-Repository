```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2, venn2_circles

# Set up the figure size
fig, ax = plt.subplots(figsize=(8, 7))

# 1. Draw the filled Venn regions with custom set sizes (Equal size circles)
v = venn2(
    subsets=(1, 1, 0.4),
    set_labels=('', ''),  # Hide default set labels to manually position custom text
    ax=ax
)

# Customize region colors and opacities 
v.get_patch_by_id('10').set_color('#FFD54F')  # Left region (Yellow) 
v.get_patch_by_id('10').set_alpha(0.7)

v.get_patch_by_id('01').set_color('#29B6F6')  # Right region (Blue)
v.get_patch_by_id('01').set_alpha(0.7)

v.get_patch_by_id('11').set_color('#4DB6AC')  # Intersection region (Teal) 
v.get_patch_by_id('11').set_alpha(0.8)

# 2. Add thick outer borders to the circles 
c = venn2_circles(subsets=(1, 1, 0.4), ax=ax) 
c[0].set_edgecolor('#FFB300')  # Darker yellow border 
c[0].set_linewidth(6) 
c[1].set_edgecolor('#03A9F4')  # Darker blue border 
c[1].set_linewidth(6) 

# 3. Add custom text labels to each region 
# Left text 
v.get_label_by_id('10').set_text("Creatures with\ntwo legs that\ncannot fly") 
v.get_label_by_id('10').set_text("Sauce\nVenn Diagram")

# add another text 
v.get_label_by_id('10').set_text("Mayonaise")
v.get_label_by_id('10').set_fontsize(11) 
v.get_label_by_id('10').set_fontweight('bold') 

# Middle text (Intersection) 
v.get_label_by_id('11').set_text("Creatures\nwith two legs\nthat can fly") 

# add another text
v.get_label_by_id('11').set_text("Special Sauce")

v.get_label_by_id('11').set_fontsize(11) 
v.get_label_by_id('11').set_fontweight('bold') 

# Right text 
v.get_label_by_id('01').set_text("Creatures that\ncan fly, with\nfewer or more\nthan two legs") 

# add another text
v.get_label_by_id('01').set_text("Ketchup")

v.get_label_by_id('01').set_fontsize(11) 
v.get_label_by_id('01').set_fontweight('bold')

# 4. Add Top Title 
plt.title(
    "Creatures that cannot\nfly, with fewer or more\nthan two legs",
    fontsize=14,
    fontweight='bold', 
    pad=20
)

# 5. Add an outer black bounding box
fig.patch.set_edgecolor('black') 
fig.patch.set_linewidth(3) 

plt.tight_layout() 
plt.show()
```

#### Listing multiple topics in the Venn-Diagram

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

fig, ax = plt.subplots(figsize=(14, 13))
v = venn2(subsets=(1, 1, 0.4), set_labels=('', ''), ax=ax)

# Define your topics as clean Python lists
bioinformatics_topics = [
    "Bioinformatics\n",
    "\u2022 High-Throughput Sequencing",
    "\u2022 Metagenomic Profiling",
    "\u2022 Variant Calling",
    "\u2022 Transposons and Retrotransposons",
    "\u2022 Epitomes",
    "\u2022 Antimicrobial Resistance Genes", 
    "\u2022 Gene Regulation",
    "\u2022 Omics: \n Gemonics-DNA\n Transcriptomics-RNA\n Proteomics-Proteome\n Metabolomics-Metabolome",
    "\u2022 Statistical Genetics", 
    "\u2022 Etiology"
    
]

intersection_topics = [
    "Urban Metagenomics\n",
    "• Built-Environment Microbes",
    "• Spatial Microbiome Tracking"
]

urbanism_topics = [
    "New Urbanism\n",
    "• Walkability",
    "• Transit-Oriented Dev",
    "• Mixed-Use Zoning"
]

# Pass joined lists into set_text()
v.get_label_by_id('10').set_text("\n".join(bioinformatics_topics))
v.get_label_by_id('11').set_text("\n".join(intersection_topics))
v.get_label_by_id('01').set_text("\n".join(urbanism_topics))

plt.show()
```
